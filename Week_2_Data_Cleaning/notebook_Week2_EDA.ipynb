# Week 2: Exploratory Data Analysis (EDA) - HR Attrition Diagnostics

This notebook executes the Exploratory Data Analysis (EDA) outlined in `HR_Attrition_EDA_Report.docx`.
We investigate class imbalance, compensation distribution, department attrition risk, and the **Loyalty Penalty** hypothesis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
%matplotlib inline
```

## 1. Data Ingestion & Baseline Summary

```python
df = pd.read_csv('../Week_1_Data_Cleaning/HR_Analytics_Cleaned.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 2. Univariate Visualizations: Class Imbalance & Age KDE

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Attrition Countplot (Class Imbalance)
sns.countplot(data=df, x='Attrition', palette=['#38bdf8', '#f43f5e'], ax=axes[0])
axes[0].set_title('Attrition Class Distribution (16.1% Attrition Rate)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Number of Employees')

# 2. Age Distribution with KDE
sns.histplot(df['Age'], kde=True, color='#818cf8', bins=20, ax=axes[1])
axes[1].set_title('Workforce Age Distribution (KDE Overlay)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Age (Years)')

plt.tight_layout()
plt.show()
```

## 3. Bivariate Analysis: Income Anomaly & Department Attrition

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. Monthly Income Boxplot by Attrition
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette=['#34d399', '#f43f5e'], ax=axes[0])
axes[0].set_title('Monthly Income by Attrition Status', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Monthly Income ($)')

# 2. Department Attrition Rate
dept_attr = df.groupby('Department')['Attrition_Numeric'].mean().reset_index()
dept_attr['AttritionRate_%'] = dept_attr['Attrition_Numeric'] * 100
sns.barplot(data=dept_attr, x='Department', y='AttritionRate_%', palette='Reds_r', ax=axes[1])
axes[1].set_title('Attrition Rate (%) by Department', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Attrition Rate (%)')

plt.tight_layout()
plt.show()
```

## 4. Multivariate Analysis & The Loyalty Penalty Hypothesis

- `TotalWorkingYears` vs `MonthlyIncome` Correlation: **0.77**
- `YearsAtCompany` vs `MonthlyIncome` Correlation: **0.51**
- **Loyalty Penalty**: Internal wage growth lags external market rates.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Correlation Matrix Heatmap
cols_heat = ['MonthlyIncome', 'TotalWorkingYears', 'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'Age', 'Attrition_Numeric']
sns.heatmap(df[cols_heat].corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=axes[0])
axes[0].set_title('Correlation Matrix Heatmap', fontsize=12, fontweight='bold')

# Loyalty Penalty Scatter Plot
sns.scatterplot(data=df, x='TotalWorkingYears', y='MonthlyIncome', hue='YearsAtCompany', palette='viridis', alpha=0.7, ax=axes[1])
axes[1].set_title('Loyalty Penalty: Total Working Yrs (0.77) vs Company Tenure (0.51)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Total Working Years (Industry Experience)')
axes[1].set_ylabel('Monthly Income ($)')

plt.tight_layout()
plt.show()
```

## 5. Strategic EDA Takeaways

1. **Class Imbalance**: High imbalance (~16.1% attrition) mandates using F1-score and PR-AUC for future predictive modeling.
2. **Loyalty Penalty**: Employees with high total working years but low company tenure earn more, proving internal pay lag.
3. **Department Risk**: Sales division exhibits the highest turnover, requiring targeted retention programs.
